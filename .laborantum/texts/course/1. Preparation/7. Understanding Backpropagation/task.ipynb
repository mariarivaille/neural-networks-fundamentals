# Task 1

Write a function that calculates the value of this graph for any given input vector `x` and set of coefficients $b_1, b_2, c_1, c_2$ packed as vector of weights `w`.
In this function also return all the $z$ values.

````python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def graph_value(x, w):
    b1, b2, c1, c2 = w
    
    z1 = float(x[0] + b1)
    z2 = float(x[1] + b2)
    z3 = float(sigmoid(z1))
    z4 = float(sigmoid(z2))
    z5 = float(np.tanh(z2))
    z6 = float(z5 * c2)
    z7 = float(z1 * z4)
    z8 = float(z7 * c1)
    z9 = float(z3 * z6)
    y = float(z8 + z9)

    z = [z1, z2, z3, z4, z5, z6, z7, z8, z9]
    return [y, z]
````

````python
answer['graph_value'] = [graph_value(**input) for input in inputs]
````

# Task 2

In this graph, find all direct derivatives of all the values over $z$ or $x$ or $w$.

For example, if $y = z_8 + z_9$, you need to find only $\partial_{z_9} y$ and $\partial_{z_8} y$ for that operation.

You have to do that for all the intermediate signals.

Write the result in the form of a dictionary, for example, $\partial_{z_9} y$:

```
['dy_dz9'] = 1
```

````python
def graph_derivatives(x, w):
    res = {}
    
    b1, b2, c1, c2 = w

    z1 = float(x[0] + b1)
    z2 = float(x[1] + b2)
    z3 = float(sigmoid(z1))
    z4 = float(sigmoid(z2))
    z5 = float(np.tanh(z2))
    z6 = float(z5 * c2)
    z7 = float(z1 * z4)
    z8 = float(z7 * c1)
    z9 = float(z3 * z6)

    res['dz1_dx1'] = 1
    res['dz1_db1'] = 1

    res['dz2_dx2'] = 1
    res['dz2_db2'] = 1

    res['dz3_dz1'] = float(z3 * (1 - z3))

    res['dz4_dz2'] = float(z4 * (1 - z4))

    res['dz5_dz2'] = float(1 - z5 ** 2)

    res['dz6_dz5'] = float(c2)
    res['dz6_dc2'] = z5

    res['dz7_dz1'] = z4
    res['dz7_dz4'] = z1

    res['dz8_dz7'] = float(c1)
    res['dz8_dc1'] = z7

    res['dz9_dz3'] = z6
    res['dz9_dz6'] = z3

    res['dy_dz8'] = 1
    res['dy_dz9'] = 1

    return res
````

````python
answer['graph_derivatives'] = [graph_derivatives(**input) for input in inputs]
````

# Task 3

Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{c_1} y$
- calculate that derivative

````python
def dy_dc1(x, w):
    selected_derivs = {}

    b1, b2, c1, c2 = w

    z1 = float(x[0] + b1)
    z2 = float(x[1] + b2)
    z3 = float(sigmoid(z1))
    z4 = float(sigmoid(z2))
    z5 = float(np.tanh(z2))
    z6 = float(z5 * c2)
    z7 = float(z1 * z4)

    selected_derivs['dy_dz8'] = 1
    selected_derivs['dz8_dc1'] = z7

    result = z7

    return [selected_derivs, result]
````

````python
answer['dy_dc1'] = [dy_dc1(**input) for input in inputs]
````

# Task 5
Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{b_1} y$
- calculate that derivative

````python
def dy_db1(x, w):
    selected_derivs = {}
    
    b1, b2, c1, c2 = w

    z1 = float(x[0] + b1)
    z2 = float(x[1] + b2)
    z3 = float(sigmoid(z1))
    z4 = float(sigmoid(z2))
    z5 = float(np.tanh(z2))
    z6 = float(z5 * c2)
    z7 = float(z1 * z4)
    z8 = float(z7 * c1)
    z9 = float(z3 * z6)

    selected_derivs['dz1_db1'] = 1
    selected_derivs['dz3_dz1'] = float(z3 * (1 - z3))
    selected_derivs['dz7_dz1'] = z4
    selected_derivs['dz9_dz3'] = z6
    selected_derivs['dz8_dz7'] = float(c1)
    selected_derivs['dy_dz8'] = 1
    selected_derivs['dy_dz9'] = 1

    result = float(c1 * z4 + z3 * (1 - z3) * z6)

    return [selected_derivs, result]
````

````python
answer['dy_db1'] = [dy_db1(**input) for input in inputs]
````

# Task 6
Using the values of the derivatives, calculated above:
- extract a dictionary of values that are needed to calculate $\partial_{b_2} y$
- calculate that derivative

````python
def dy_db2(x, w):
    selected_derivs = {}
    
    b1, b2, c1, c2 = w

    z1 = float(x[0] + b1)
    z2 = float(x[1] + b2)
    z3 = float(sigmoid(z1))
    z4 = float(sigmoid(z2))
    z5 = float(np.tanh(z2))
    z6 = float(z5 * c2)
    z7 = float(z1 * z4)
    z8 = float(z7 * c1)
    z9 = float(z3 * z6)

    selected_derivs['dz2_db2'] = 1
    selected_derivs['dz4_dz2'] = float(z4 * (1 - z4))
    selected_derivs['dz5_dz2'] = float(1 - z5 ** 2)
    selected_derivs['dz6_dz5'] = float(c2)
    selected_derivs['dz7_dz4'] = z1
    selected_derivs['dz8_dz7'] = float(c1)
    selected_derivs['dz9_dz6'] = z3
    selected_derivs['dy_dz8'] = 1
    selected_derivs['dy_dz9'] = 1

    dz4_dz2 = z4 * (1 - z4)
    dz5_dz2 = 1 - z5 ** 2
    result = float(c1 * z1 * dz4_dz2 + z3 * c2 * dz5_dz2)

    return [selected_derivs, result]
````

````python
answer['dy_db2'] = [dy_db2(**input) for input in inputs]
````
